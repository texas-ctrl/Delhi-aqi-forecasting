```python
import os
import pandas as pd

file_path = "../processed_data/modeling_data.pkl"

final = pd.read_pickle(file_path)

print("Loaded:", os.path.exists(file_path))
print("Shape:", final.shape)
```

```python
final.columns.tolist()
```

```python
# Keep only winter months
winter_model = final[
    final.index.month.isin([11, 12, 1, 2])
].copy()

# Remove rows where the 48-hour target cannot be calculated
winter_model = winter_model[
    winter_model['AQI_change_48h'].notna()
].copy()

# Target
y = winter_model['target_48h']

# Remove future-looking and target columns
drop_columns = [
    'AQI_24h_ahead',
    'AQI_48h_ahead',
    'AQI_72h_ahead',
    'AQI_change_48h',
    'AQI_change_72h',
    'target_48h',
    'target_72h'
]

X = winter_model.drop(columns=drop_columns)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Number of features:", X.shape[1])
print("Missing feature values:", X.isna().sum().sum())
```

```python
missing = X.isna().sum()

print(missing[missing > 0].sort_values(ascending=False))
```
